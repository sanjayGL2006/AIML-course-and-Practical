```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import joblib
# Data
height = [[150], [155], [160], [165], [170],
          [175], [180], [185], [190], [195]]

weight = [50, 53, 56, 60, 65, 70, 75, 80, 85, 90]
# 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    height, weight, test_size=0.2, random_state=42
)
# Create model
model = LinearRegression()
# Train model
model.fit(X_train, y_train)
# Predict
prediction = model.predict([[155]])
print(X_train)
print(y_train)
print(X_test)
print(y_test)
print("Predicted weight:", prediction[0])

# Save model

joblib.dump(model, "height_weight_model.pkl")

print("Model saved successfully!")
```
