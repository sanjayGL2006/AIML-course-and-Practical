```python
import pandas as pd

# ----------------------------------------------------
# Step 1: Load the Dataset
# ----------------------------------------------------
# Update filename/path if loading CSV or Excel
file_path = "/content/Sample - Superstore.csv"  # or "Sample - Superstore.xlsx"
df = pd.read_csv(file_path, encoding="latin1")

# ----------------------------------------------------
# Step 2: Understand the Dataset
# ----------------------------------------------------
print("--- 1. Shape of Dataset (Rows, Columns) ---")
print(df.shape)

print("\n--- 2. Number of Columns ---")
print(len(df.columns))

print("\n--- 3. Column Names ---")
print(df.columns.tolist())

print("\n--- 4. Data Types ---")
print(df.dtypes)

print("\n--- 5. First 5 Rows ---")
print(df.head(5))

print("\n--- 6. Last 5 Rows ---")
print(df.tail(5))

print("\n--- 7. Random 10 Records ---")
print(df.sample(10))

# ----------------------------------------------------
# Step 3: Data Selection & Filtering
# ----------------------------------------------------
# 1. Select specific columns
selected_cols = df[["Order ID", "Customer Name", "City", "Sales", "Profit"]]
print("\n--- Selected Columns (Preview) ---")
print(selected_cols.head())

# 2. Select rows by index (rows 10 to 20)
rows_10_to_20 = df.iloc[10:21]
print("\n--- Rows from Index 10 to 20 ---")
print(rows_10_to_20)

# ----------------------------------------------------
# Step 4: Handling Missing Values & Duplicates
# ----------------------------------------------------
print("\n--- Count Missing Values per Column ---")
print(df.isnull().sum())

# Fill missing values in specific columns
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

if "Quantity" in df.columns:
    df["Quantity"] = df["Quantity"].fillna(0)

# Fill other numerical missing values with the column mean
numeric_cols = df.select_dtypes(include=["float64", "int64"]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

# Count and remove duplicate records
duplicate_count = df.duplicated().sum()
print(f"\n--- Total Duplicate Records: {duplicate_count} ---")
df = df.drop_duplicates()

# ----------------------------------------------------
# Step 5: Data Transformations & Cleaning
# ----------------------------------------------------
# Convert text columns to Upper / Lower case
if "Country" in df.columns:
    df["Country"] = df["Country"].astype(str).str.upper()

if "City" in df.columns:
    df["City"] = df["City"].astype(str).str.lower()

# Convert Date columns to DateTime format
if "Order Date" in df.columns:
    df["Order Date"] = pd.to_datetime(df["Order Date"])

if "Ship Date" in df.columns:
    df["Ship Date"] = pd.to_datetime(df["Ship Date"])

# ----------------------------------------------------
# Step 6: Save the Cleaned Dataset
# ----------------------------------------------------
output_path = "Sample_Superstore_Cleaned.csv"
df.to_csv(output_path, index=False)
print(f"\nCleaned dataset successfully saved to: {output_path}")
```
