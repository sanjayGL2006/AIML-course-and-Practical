```python
import matplotlib.pyplot as plt
x=np.arange(6)
y=np.arange(0,11,2)
print(x,y)
plt.plot(x,y,marker='s',ms=12,mec='r',mfc='y',ls='dashed',lw=10,color="cyan")
plt.grid()
f1={'family':'serif','color':'red','size':15}
f2={'family':'calibri','color':'blue','size':15}
f3={'family':'serif','color':'green','size':22}
plt.title('Raza Prithvi',fontdict=f3,loc='center')
plt.xlabel('Raza',fontdict=f1)
plt.ylabel('Prithvi',fontdict=f2)
plt.show()
```

```python
year=[2012,2014,2016,2018,2020]
Bharat =[100.34,96.0,200,394.5,795.60]
Nepal =[10.1,20,30.4,50.6,60.9]
plt.plot(year,Bharat,color='orange',label='Bharat')
plt.plot(year,Nepal,color='blue',label='Nepal')
plt.xlabel('years')
plt.ylabel('power consumption in kwh')
plt.title('Electricity consumption per capita in Bharat and Nepal')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0.0,2.0,0.01)
y=1 + np.sin(2 * np.pi * x)

fig, ((ax1,ax2),(ax3,ax4),(ax5,ax6),(ax7,ax8),(ax9,ax10))=plt.subplots(5,2)
ax1.plot(x,y,color='red')
ax2.plot(x,y,color='orange')
ax3.plot(x,y,color='green')
ax4.plot(x,y,color='cyan')
ax5.plot(x,y,color='hotpink')
ax6.plot(x,y,color='blue')
ax7.plot(x,y,color='black')
ax8.plot(x,y,color='yellow')
ax9.plot(x,y,color='violet')
ax10.plot(x,y,color='brown')
```

```python
continents =['asia','Europe','north america','south america','australia','africa','antatica']
area =[25,20,15,10,15,10,5]
explode =(0,0.1,0.1,0.1,0.1,0,0)
plt.pie(area,explode=explode,labels=continents,autopct='%1.1f%%',startangle=0,wedgeprops={"edgecolor": "black", 'linewidth': 2, 'antialiased': True})
plt.axis('equal')
plt.show()
```
