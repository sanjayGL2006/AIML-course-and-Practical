```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(6)
y=np.arange(0,11,2)
print(x,y)
plt.plot(x,y,marker='o',ms=12,mec='r',mfc='y',ls='dashed',lw=10,color='green')
plt.grid()
f1={'family':'serif','color':'red','size':15}
f2={'family':'serif','color':'blue','size':15}
plt.title('Sample plot',fontdict=f2,loc='right')
plt.xlabel('X-Axis',fontdict=f2)
plt.ylabel('Y-Axis',fontdict=f1)
plt.show()
```

```python
year = [2012,2014,2016,2018,2020]
India = [100.34,96.0,200,394.5,795.6]
Nepal = [10.1,20,30.4,50.6,60.9]
plt.plot(year,India,color='orange',label='India')
plt.plot(year,Nepal,color='red',label='Nepal')
plt.xlabel('years')
plt.ylabel('power consumption is kwh')
plt.title('Electricity consumption per capita in India and Nepal')
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x=np.arange(0.0,2.0,0.01)
y=1 + np.sin(2 * np.pi * x)

fig, ((ax1,ax2),(ax3,ax4),(ax5,ax6),(ax7,ax8),(ax9,ax10))=plt.subplots(5,2)
ax1.plot(x,y,color='red')
ax2.plot(x,y,color='orange')
ax3.plot(x,y,color='green')
ax4.plot(x,y,color='cyan')
ax5.plot(x,y,color='hotpink')
ax6.plot(x,y,color='blue')
ax7.plot(x,y,color='black')
ax8.plot(x,y,color='yellow')
ax9.plot(x,y,color='violet')
ax10.plot(x,y,color='brown')
```

```python
a=np.array([[9,1,2,7,5,8,3,4,6],
                [4,2,3,7,9,1,6,5,8]])
categories=np.array([0,1,2,0,1,2,0,1,2])
colormap=np.array(['r','g','k'])
plt.scatter(a[0],a[1], s=100, c=colormap[categories])
plt.title('sample pllot')
plt.xlabel('X-Axis')
plt.ylabel('Y-Axis')
plt.show()
```

```python
x=["A","B","C","D"]
y=[1,2,3,4]
plt.bar(x,y)

for index, value in enumerate(y):
  plt.text(value,index,
           str(value))
  plt.show()
```

```python
continents=['Asia','Europe','N America','S America','Australia','Africa','Antarctica']
area=[25,20,15,10,5]
explode=(0.1,0,0.1,0,0.1,0.1,0.1)
plt.pie(area,explode=explode,labels=continents,autopct='%1.1f%%',startangle=0,wedgeprops={"edgecolor":"black",'linewidth':2,'antialiased':True})
plt.axis('equal')
plt.show()
```
